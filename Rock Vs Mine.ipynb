# Import dependencises


```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df_sona_data = pd.read_csv(r"D:\Data_Science\5), Machine_Learning\1st_ML_Submarine_prediction\sona_data.csv", encoding="latin-1",header=None)
```

## Seperate data 

```python
x = df_sona_data.drop(columns= 60,axis = 1)
y = df_sona_data[60]
```

```python
print(x)
```

```python
print(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size= 0.1 ,stratify = y,  random_state= 42)
```

```python
X_train  ## the 80 % of data for trainning
```

```python
X_test ## 20 % for testing 
```

```python
print(x.shape , X_train.shape,X_test.shape)
```

```python
# create the object model
## traing model
model = LogisticRegression()
model.fit(X_train,y_train)
```

# Model Evaluation

```python
# accuray on training data
X_train_prediction = model.predict(X_train)
```

```python
training_data_accuracy = accuracy_score(X_train_prediction, y_train)
```

```python
print('Accuracy on training data',training_data_accuracy )  
```

```python
# accuracy on test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction,y_test)
```

```python
print('Accuracy on test data',test_data_accuracy)
```
